```python
import heapq

class Node:
  def __init__(self, name, parent=None, g=0, h=0):
    self.name=name
    self.parent=parent
    self.g=g  #cost from start to this node
    self.h=h  #Heuristic estimate to goal
    self.f=g+h  #total estimated cost

  def __lt__(self, other):
    return self.f < other.f

def a_star_search(graph, heuristics, start, goal):
  open_list = []  #Priority queue for A*
  closed_list = set()

  start_node = Node(start, None, 0, heuristics[start])
  heapq.heappush(open_list, start_node)

  while open_list:
    current_node = heapq.heappop(open_list)

    if current_node.name == goal:
      path = []
      while current_node:
        path.append(current_node.name)
        current_node = current_node.parent
      return path[::-1] #Return reversed path

    closed_list.add(current_node.name)

    for neighbor, cost in graph[current_node.name].items():
      if neighbor in closed_list:
        continue

      g_cost = current_node.g + cost
      h_cost = heuristics[neighbor]
      new_node = Node(neighbor, current_node, g_cost, h_cost)

      heapq.heappush(open_list, new_node)

  return None #No path found

#Example graph representation as an adjacency list
graph = {
    'A': {'B': 1, 'C': 4},
    'B': {'A': 1, 'D': 2, 'E': 5},
    'C': {'A': 4, 'F': 3},
    'D': {'B': 2, 'E': 1, 'G': 3},
    'E': {'B': 5, 'D': 1, 'H': 2},
    'F': {'C': 3, 'I': 6},
    'G': {'D': 3, 'H': 2},
    'H': {'E': 2, 'G': 2, 'I': 1},
    'I': {'F': 6, 'H': 1}
}

#Heuristic values (straight-line distances to goal 'I')
heuristics = {
    'A': 10, 'B': 8, 'C': 7, 'D': 6, 'E': 4, 'F': 3,
    'G': 4, 'H': 2, 'I': 0
}

start = 'A'
goal = 'I'

path = a_star_search(graph, heuristics, start, goal)

if path:
  print(f"Shortest path from {start} to {goal}: {path}")
else:
  print(f"No path found from {start} to {goal}")
```

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
```

```python
#Load dataset
data = load_iris()
x = data.data  # Change from data.data() to data.data
y = data.target
features = list(range(x.shape[1]))  #Feature indices

#Heuristic function: Estimate importance based on single-feature
def heuristic_feature_score(feature):
  x_train, x_test, y_train, y_test = train_test_split(x[:, feature].reshape(-1, 1), y, test_size=0.2, random_state=42)
  model = RandomForestClassifier(random_state=42)
  model.fit(x_train, y_train)
  return accuracy_score(y_test, model.predict(x_test))

#Rank all features by their heuristic scores
feature_scores = {f: heuristic_feature_score(f) for f in features}
sorted_features = sorted(feature_scores.items(), key=lambda x: x[1], reverse=True)

#Dispaly results
print("Feature Rankings (Feature Index, Accuracy Score):")
for feature, score in sorted_features:
  print(f"Feature {feature}: {score:.4f}")

#Best feature
best_feature = sorted_features[0][0]
print("\nBest Feature Selected:",best_feature)
```
